```python
field_size = (10, 10)

field = np.random.binomial(1, 0.2, size=field_size)

def step(field):
    neigh_cnt = np.zeros_like(field)
    for i in range(2):
        for j in range(2):
            field_shifted = np.roll(field, (i, j))
            neigh_cnt += field_shifted
            
            
    new_field = ((neigh_cnt == 2) | (neigh_cnt == 3)).astype(np.int64)

    return new_field
```

```python
import time

for _ in range(10):
    print(field)
    field = step(field)
    print("\n"*10)
    time.sleep(2)
```
