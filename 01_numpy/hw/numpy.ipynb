Ниже приведён вспомогательный код. В нём циклы разрешены. При необходимости уменьшите размеры данных или число повторений.

````python
def count_time(func, *args, repeats=5, number=10):
    """Вернуть медианное время одного вызова в миллисекундах."""
    if repeats < 1 or number < 1:
        raise ValueError("repeats и number должны быть положительными")
    func(*args)  # Прогрев перед измерениями.
    measurements = []
    for _ in range(repeats):
        start = time.perf_counter()
        for _ in range(number):
            func(*args)
        measurements.append((time.perf_counter() - start) / number)
    return float(np.median(measurements)) * 1000
````
